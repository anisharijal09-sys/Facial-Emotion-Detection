```python
import os
from IPython.display import Image as IPyImage, display

# Base folder containing YOLO runs
runs_dir = '/content/Facial-Expression-1/runs/detect/'

# List all run folders starting with 'train'
run_folders = [f for f in os.listdir(runs_dir) if f.startswith('train') and os.path.isdir(os.path.join(runs_dir, f))]

if run_folders:
    # Pick the latest run folder
    latest_folder = max(run_folders, key=lambda f: os.path.getmtime(os.path.join(runs_dir, f)))
    latest_path = os.path.join(runs_dir, latest_folder)
    print("Latest run folder:", latest_path)

    # Only get validation prediction images (val_batch*_pred.jpg)
    val_pred_images = sorted([img for img in os.listdir(latest_path) if img.startswith('val_batch') and img.endswith('_pred.jpg')])

    # Display the first 2 or 3 images
    for img in val_pred_images[:3]:
        display(IPyImage(filename=os.path.join(latest_path, img), width=600))
else:
    print("No YOLO run folders found in:", runs_dir)
```
